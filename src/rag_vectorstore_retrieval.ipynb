# Documents and Embeddings

```python
from langchain_community.document_loaders import PyPDFDirectoryLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma

# Loading the documents
loader = PyPDFDirectoryLoader("../corpus/")
documents = loader.load()
print(f" {len(documents)} total pages.")

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 500,
    chunk_overlap = 50
)

# Split the text
chunks = text_splitter.split_documents(documents)
print(f"Documents have been splitted into {len(chunks)} chunks.")

print("Starting embeddings")
embeddings_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

# Chroma index
print("Creating and saving vectorial index")
vector_store = Chroma.from_documents(
    documents=chunks,
    embedding=embeddings_model,
    persist_directory="./chroma_db"
)
```

# RETRIEVAL TESTS

```python
query_1 = "What is the Attention mechanism? "
query_2 = "What are the advantages of retrieval-augmented generation (RAG)?"

print(f"\n--- Results for: '{query_1}' ---")
results_1 = vector_store.similarity_search(query_1,k=2)
for i,res in enumerate(results_1):
    print(f"\nResultado {i+1}:")
    print(res.page_content)

print(f"\n--- Results for: '{query_2}' ---")
results_2 = vector_store.similarity_search(query_2,k=2)
for i,res in enumerate(results_2):
    print(f"\nResultado {i+1}:")
    print(res.page_content)
```
